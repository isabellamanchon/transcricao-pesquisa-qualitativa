```python
from google import genai
from google.colab import userdata

client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))

caminho = "/content/transcrisa.txt"
try:
    with open(caminho, encoding="utf-8") as f:
        transcricao = f.read()
except UnicodeDecodeError:
    with open(caminho, encoding="cp1252") as f:
        transcricao = f.read()

PROMPT = """Você é um analista de discurso político. Abaixo está a transcrição
de um debate, com horário e nome do falante em cada linha.
Para cada falante, entregue:
1. Os 3 temas principais abordados
2. As propostas concretas apresentadas (apenas as que aparecem no texto)
3. Números ou dados citados, com o trecho
4. Uma pergunta que ficou sem resposta, se houver
Responda em português, em markdown, sem inventar nada que não esteja na transcrição."""

resposta = client.models.generate_content(
    model="gemini-3-flash-preview",
    contents=PROMPT + "\n\nTRANSCRIÇÃO:\n" + transcricao,
)

resumo = resposta.text
print(resumo)
open("/content/resumo_debate.md", "w", encoding="utf-8").write(resumo)
```
